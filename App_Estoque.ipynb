```python

produtos = ["Teclado", "Mouse", "Monitor"]

opcao = ""

while opcao != "0":

    print("\n=== ESTOQUE FÁCIL ===")
    print("1 - Listar produtos")
    print("2 - Cadastrar produto")
    print("3 - Atualizar produto")
    print("4 - Excluir produto")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")

    match opcao:

        # LISTAR PRODUTOS
        case "1":
            print("\n=== PRODUTOS CADASTRADOS ===")

            for produto in produtos:
                print(produto)

        # CADASTRAR PRODUTO
        case "2":
            novo_produto = input("Digite o nome do produto: ")

            produtos.append(novo_produto)

            print("Produto cadastrado com sucesso!")

        # ATUALIZAR PRODUTO
        case "3":
            produto_antigo = input(
                "Qual produto deseja atualizar? "
            )

            if produto_antigo in produtos:
                novo_produto = input("Digite o novo nome: ")

                produtos.remove(produto_antigo)
                produtos.append(novo_produto)

                print("Produto atualizado com sucesso!")
            else:
                print("Produto não encontrado!")

        # EXCLUIR PRODUTO
        case "4":
            produto_excluir = input(
                "Qual produto deseja excluir? "
            )

            if produto_excluir in produtos:
                produtos.remove(produto_excluir)

                print("Produto excluído com sucesso!")
            else:
                print("Produto não encontrado!")

        # SAIR DO SISTEMA
        case "0":
            print("Encerrando o sistema...")

        # OPÇÃO INVÁLIDA
        case _:
            print("Opção inválida!")

print("Programa finalizado.")
```
