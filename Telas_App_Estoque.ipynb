```python

# ============================================
# ESTOQUE FÁCIL
# Sistema de controle de produtos com interface
# Biblioteca: ipywidgets
# ============================================

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# --------------------------------------------
# 1. DADOS DO SISTEMA
# --------------------------------------------

produtos = ["Teclado", "Mouse", "Monitor"]


# --------------------------------------------
# 2. CAMPOS DA INTERFACE
# --------------------------------------------

campo_cadastro = widgets.Text(
    description="Produto:",
    placeholder="Digite o nome do produto",
    layout=widgets.Layout(width="350px")
)

campo_antigo = widgets.Text(
    description="Atual:",
    placeholder="Nome atual do produto",
    layout=widgets.Layout(width="350px")
)

campo_novo = widgets.Text(
    description="Novo:",
    placeholder="Digite o novo nome",
    layout=widgets.Layout(width="350px")
)

campo_excluir = widgets.Text(
    description="Produto:",
    placeholder="Nome do produto para excluir",
    layout=widgets.Layout(width="350px")
)


# --------------------------------------------
# 3. BOTÕES
# --------------------------------------------

botao_listar = widgets.Button(
    description="Listar produtos",
    button_style="info",
    icon="list"
)

botao_cadastrar = widgets.Button(
    description="Cadastrar produto",
    button_style="success",
    icon="plus"
)

botao_atualizar = widgets.Button(
    description="Atualizar produto",
    button_style="warning",
    icon="edit"
)

botao_excluir = widgets.Button(
    description="Excluir produto",
    button_style="danger",
    icon="trash"
)


# Área para mostrar mensagens e resultados
saida = widgets.Output()


# --------------------------------------------
# 4. FUNÇÕES DO SISTEMA
# --------------------------------------------

def listar_produtos():
    """Exibe os produtos cadastrados."""

    with saida:
        clear_output(wait=True)

        print("=== PRODUTOS CADASTRADOS ===")

        if produtos:
            for produto in produtos:
                print(f"- {produto}")

            print(f"\nTotal de produtos: {len(produtos)}")
        else:
            print("Nenhum produto cadastrado.")


def cadastrar_produto(nome):
    """Adiciona um produto à lista."""

    nome = nome.strip()

    if nome == "":
        return "Digite o nome do produto."

    if nome in produtos:
        return "Esse produto já está cadastrado."

    produtos.append(nome)
    return f"Produto '{nome}' cadastrado com sucesso!"


def atualizar_produto(antigo, novo):
    """Atualiza o nome de um produto existente."""

    antigo = antigo.strip()
    novo = novo.strip()

    if antigo == "" or novo == "":
        return "Preencha os dois campos."

    if antigo not in produtos:
        return "Produto não encontrado."

    if novo in produtos and novo != antigo:
        return "Já existe um produto com esse nome."

    indice = produtos.index(antigo)
    produtos[indice] = novo

    return f"Produto atualizado: '{antigo}' → '{novo}'."


def excluir_produto(nome):
    """Remove um produto da lista."""

    nome = nome.strip()

    if nome == "":
        return "Digite o nome do produto."

    if nome not in produtos:
        return "Produto não encontrado."

    produtos.remove(nome)
    return f"Produto '{nome}' excluído com sucesso."


# --------------------------------------------
# 5. AÇÕES DOS BOTÕES
# --------------------------------------------

def ao_clicar_listar(botao):
    listar_produtos()


def ao_clicar_cadastrar(botao):
    mensagem = cadastrar_produto(campo_cadastro.value)

    with saida:
        clear_output(wait=True)
        print(mensagem)

    campo_cadastro.value = ""


def ao_clicar_atualizar(botao):
    mensagem = atualizar_produto(
        campo_antigo.value,
        campo_novo.value
    )

    with saida:
        clear_output(wait=True)
        print(mensagem)

    campo_antigo.value = ""
    campo_novo.value = ""


def ao_clicar_excluir(botao):
    mensagem = excluir_produto(campo_excluir.value)

    with saida:
        clear_output(wait=True)
        print(mensagem)

    campo_excluir.value = ""


# --------------------------------------------
# 6. CONECTAR OS BOTÕES ÀS FUNÇÕES
# --------------------------------------------

botao_listar.on_click(ao_clicar_listar)
botao_cadastrar.on_click(ao_clicar_cadastrar)
botao_atualizar.on_click(ao_clicar_atualizar)
botao_excluir.on_click(ao_clicar_excluir)


# --------------------------------------------
# 7. MONTAR A INTERFACE
# --------------------------------------------

titulo = widgets.HTML(
    value="""
    <h1 style="color:#2457A7;">Estoque Fácil</h1>
    <p>Sistema de controle de produtos</p>
    <hr>
    """
)

secao_cadastro = widgets.VBox([
    widgets.HTML("<h3>Cadastrar produto</h3>"),
    campo_cadastro,
    botao_cadastrar
])

secao_atualizacao = widgets.VBox([
    widgets.HTML("<h3>Atualizar produto</h3>"),
    campo_antigo,
    campo_novo,
    botao_atualizar
])

secao_exclusao = widgets.VBox([
    widgets.HTML("<h3>Excluir produto</h3>"),
    campo_excluir,
    botao_excluir
])

secao_listagem = widgets.VBox([
    widgets.HTML("<h3>Consultar estoque</h3>"),
    botao_listar
])

interface = widgets.VBox([
    titulo,
    secao_cadastro,
    widgets.HTML("<hr>"),
    secao_atualizacao,
    widgets.HTML("<hr>"),
    secao_exclusao,
    widgets.HTML("<hr>"),
    secao_listagem,
    widgets.HTML("<hr>"),
    widgets.HTML("<h3>Resultado</h3>"),
    saida
])

# Exibir a interface
display(interface)

# Mostrar os produtos iniciais
listar_produtos()
```
